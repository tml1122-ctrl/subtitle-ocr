```python
#@title 字幕函數
def parse_srt(srt_path):
    with open(srt_path, "r", encoding="utf-8") as f:
        content = f.read()

    # 每段字幕區塊以空行分隔
    blocks = re.split(r"\n\s*\n", content.strip())
    subs = []

    for block in blocks:
        lines = block.strip().splitlines()
        if len(lines) < 2:
            continue

        # 嘗試抓時間軸
        match = re.search(
            r"(\d{2}):(\d{2}):(\d{2}),(\d{3})\s*-->\s*(\d{2}):(\d{2}):(\d{2}),(\d{3})",
            lines[1]
        )
        if not match:
            continue

        h1, m1, s1, ms1, h2, m2, s2, ms2 = map(int, match.groups())
        start = h1 * 3600 + m1 * 60 + s1 + ms1 / 1000
        end = h2 * 3600 + m2 * 60 + s2 + ms2 / 1000

        subs.append({
            "index": int(lines[0]) if lines[0].isdigit() else len(subs) + 1,
            "start": start,
            "end": end
        })

    print(f"✅ 已解析 {len(subs)} 段字幕")
    return subs

'''
subs = parse_srt("/content/sample.srt")
'''


def replace_srt_text(srt_path, ocr_results, output_path):
    subs = pysrt.open(srt_path, encoding='utf-8')

    for sub in subs:
        start_sec = (
            sub.start.hours * 3600
            + sub.start.minutes * 60
            + sub.start.seconds
            + sub.start.milliseconds / 1000
        )

        # 找到最接近的 OCR 結果（誤差1秒內）
        best_match = min(
            ocr_results,
            key=lambda r: abs(r[0] - start_sec),
            default=None
        )

        if best_match and abs(best_match[0] - start_sec) < 1:
            sub.text = best_match[1]

    #subs.save(output_path, encoding='utf-8')
    subs.save(output_path, encoding='utf-8-sig')
    print(f"✅ 已儲存覆蓋字幕：{output_path}")
```

```python
#@title OCR
def ocr_by_srt_with_json(video_path, srt_path, json_path, ocr, max_shift=1):
    subs = parse_srt(srt_path)
    with open(json_path, "r", encoding="utf-8") as f:
        boxes_data = json.load(f)

    cap = cv2.VideoCapture(video_path)
    fps = cap.get(cv2.CAP_PROP_FPS)
    total_frames = int(cap.get(cv2.CAP_PROP_FRAME_COUNT))
    print(f"🎞️ FPS: {fps:.2f}, Frame count: {total_frames}")

    # 建立時間索引，方便搜尋最接近的 JSON entry
    json_times = [entry["time"] for entry in boxes_data]

    results = []
    start_time = time.time()

    for sub in tqdm(subs, desc="🎬 OCR Running"):
        t = sub["start"]
        # 找最接近的 JSON entry
        idx = bisect.bisect_left(json_times, t)
        if idx >= len(boxes_data):
            continue

        # 嘗試 OCR
        found_text = False
        for shift in range(max_shift + 1):
            curr_idx = idx + shift
            if curr_idx >= len(boxes_data):
                break

            entry = boxes_data[curr_idx]
            boxes = entry["boxes"]
            if not boxes:
                continue

            boxes_sorted = sorted(boxes, key=lambda b: b[1])

            frame_idx = int(entry["time"] * fps)
            if frame_idx >= total_frames:
                continue
            cap.set(cv2.CAP_PROP_POS_FRAMES, frame_idx)
            ret, frame = cap.read()
            if not ret:
                continue

            all_texts = []
            for (x1, y1, x2, y2) in boxes_sorted:
                roi_img = frame[y1:y2, x1:x2]
                ocr_result = ocr.predict(roi_img)
                for res in ocr_result:
                    if "rec_texts" in res:
                        all_texts.extend([txt for txt in res["rec_texts"] if txt.strip()])

            if all_texts:
                line = "\n".join(all_texts)
                results.append((sub["start"], line))
                print(f"[{sub['index']:04d}] {line}")
                found_text = True
                break

    cap.release()
    total_time = time.time() - start_time
    print(f"✅ 完成 {len(subs)} 句字幕，共耗時 {total_time:.2f} 秒")
    return results


# +++++ RUN +++++
ocr_results = ocr_by_srt_with_json(video_path, pre_srt_path, json_path, ocr)
replace_srt_text(pre_srt_path, ocr_results, "CC.srt")
```
